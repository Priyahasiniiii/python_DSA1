```python
#string manipulation
x="python is a programming language"
count=0
for i in x:
  if i in "aeiou":
    count+=1
print(count)
```

```python
x="python is a programming language"
count_vowels=0
count_consonants=0
count_spaces=0
count_special_char=0
for i in x:
  if i.isalpha():
    if i in "aeiou":
      count_vowels+=1
    else:
      count_consonants+=1
  elif i.isspace():
    count_spaces+=1
  else:
    count_special_char+=1
print(count_vowels,count_consonants,count_spaces,count_special_char)
```

```python
# Remove duplicates from string
x=input("Enter a string:")
seen=set()
result=""
for i in x:
  if i in seen:
    continue
  seen.add(i)
  result+=i
print(result)
```

```python
#reverse the string
x=input("Enter a string:")
reverse=""
for i in x:
  reverse=i+reverse
print(reverse)



```

```python
#Anagram
x=input("Enter a string:")
y=input("Enter a string:")
print(sorted(x)==sorted(y))


```

```python
x=input("Enter a string:")
y=input("Enter a string:")
if len(x)==len(y):
  print(True)
freq={}
for ch in x:
  freq[ch]=freq.get(ch,0)+1
for ch in y:
  if ch not in freq:
    print(False)
    break
  else:
    freq[ch]=-1


```

```python
#Recursion
#Factorial
def factorial(n):
  if n <= 1:
    return 1
  return n*factorial(n-1)
factorial(5)
```

```python
#Fibonacci
def Fibonacci(n):
  if n<=1:
    return n
  return Fibonacci(n-1)+Fibonacci(n-2)
for i in range(5):
  print(Fibonacci(i),end="")
```
